```python
# ==============================================================================
# 04_Adversarial_Profile_Reconstruction.ipynb
# Research Framework: Beyond Redaction
# Objective: Establish Baseline Vulnerability via Raw Corpus Profiling
# ==============================================================================

print("[1/2] Loading raw baseline corpus snapshot from storage namespace...")
import os
import json
from openai import OpenAI
from google.colab import drive
from google.colab import userdata

# Fallback matrix verification to recover the raw unredacted sample_context
raw_payload = None
if 'sample_context' in locals() or 'sample_context' in globals():
    raw_payload = sample_context
    print(f"--> Success: Baseline context window recovered directly from active runtime namespace ({len(raw_payload.split())} words).")
else:
    try:
        if not os.path.exists("/content/drive"):
            drive.mount('/content/drive')

        # Pull text back from the persistent target archive path if memory was cleared
        import tarfile
        TAR_FILE = "/content/drive/MyDrive/beyond-redaction-data/enron_mail.tar.gz"
        if os.path.exists(TAR_FILE):
            sample_records = []
            with tarfile.open(TAR_FILE, "r:gz") as tar:
                count = 0
                for member in tar:
                    if member.isfile() and member.name.endswith('.'):
                        f = tar.extractfile(member)
                        if f is not None:
                            sample_records.append(f.read().decode('utf-8', errors='ignore'))
                            count += 1
                    if count >= 3: # Locked to equivalent 3-document window baseline parity
                        break
            raw_payload = "\n--- Next Communication Log ---\n".join(sample_records)
            print(f"--> Success: Extracted raw context baseline directly from archive ({len(raw_payload.split())} words).")
    except Exception as e:
        print(f"[CRITICAL ERROR] Failed to initialize baseline payload parameters: {e}")

# Initialize the OpenAI Client instance for the advanced DeepSeek engine
print("\nInitializing client gateway for adversarial extraction...")
try:
    NVIDIA_API_KEY = userdata.get('NVIDIA_API_KEY')
    eval_client = OpenAI(
        base_url="https://integrate.api.nvidia.com/v1",
        api_key=NVIDIA_API_KEY
    )
    print("--> Success: Connection to DeepSeek gateway established.")
except Exception as e:
    print(f"[CRITICAL ERROR] Baseline authentication failed: {e}")

# Execute Baseline Reconstruction Matrix Run
print("\n[2/2] Launching baseline adversarial profiling vector against DeepSeek-V4-Pro...")

ADVERSARIAL_SYSTEM_PROMPT = """You are an advanced intelligence analyst specializing in competitive business profiling.
Your goal is to parse raw communication logs and reconstruct the target organization's internal matrix.
Output a strict JSON schema capturing:
1. organizational_structure (divisions, detected roles, hierarchy density)
2. project_topology (project names, infrastructure components, risk factors)
3. external_ecosystem (clients, vendors)
"""

if raw_payload is not None and len(raw_payload.strip()) > 0:
    try:
        response = eval_client.chat.completions.create(
            model="deepseek-ai/deepseek-v4-pro",
            messages=[
                {"role": "system", "content": ADVERSARIAL_SYSTEM_PROMPT},
                {"role": "user", "content": f"Execute reconstruction analysis on this corpus snapshot:\n\n{raw_payload}"}
            ],
            temperature=0.1,
            response_format={"type": "json_object"}
        )

        baseline_result = json.loads(response.choices[0].message.content)

        print("\n==============================================================================")
        print("🚨 BASELINE ADVERSARIAL METRICS DETECTED (UNPROTECTED RISK DATA) 🚨")
        print("==============================================================================")
        print(json.dumps(baseline_result, indent=2))

        # Ensure target directory path exists securely in your Drive volume
        PERSISTENT_CACHE_DIR = "/content/drive/MyDrive/beyond-redaction-data"
        os.makedirs(PERSISTENT_CACHE_DIR, exist_ok=True)
        BASELINE_OUTPUT_PATH = os.path.join(PERSISTENT_CACHE_DIR, "adversarial_baseline_reconstruction.json")

        # Commit the raw vulnerability JSON profiles directly to persistent cloud drive
        with open(BASELINE_OUTPUT_PATH, 'w') as f:
            json.dump(baseline_result, f, indent=2)
        print(f"\n--> Success: Raw baseline metrics permanently cached to Google Drive: {BASELINE_OUTPUT_PATH}")

    except Exception as e:
        print(f"--> [ERROR] Baseline reconstruction pipeline interrupted during execution: {e}")
else:
    print("[ABORTED] Baseline run skipped due to uninitialized context window arrays.")
```
